### Basic Terminologies of NLP:-
1. Corpus >> refers to the total text
2. Document >> Each statement/sentence in a dataset is called as a document.
3. Words >> Sequence of characters combinedly forms a word which is meaningful.
4. Vocabulary
5. Vector
6. Stopwords
7. Tokens

## Tesxt Processing Steps:-
### 1. Tokenization :- dividing/spliting of a corpus into documents or documents into words.

```python
import nltk
```

```python
doc="We are learning NLP. NLP is a amazing concept of AIML."
doc
```

```python
nltk.download('punkt_tab')
```

```python
from nltk import sent_tokenize
sent_tokenize(doc)
```

```python
from nltk import word_tokenize
word_tokenize(doc)
```

### 2. Stemming :- is a processof removing the suffixes of a word into its root form by using the nearest root word by spelling.
Example:
1. "playing" >> "play"
2. "learning">> "learn"
3. "studied" >> "study"

```python
from nltk import PorterStemmer
ps=PorterStemmer()
```

```python
ps.stem('playing')
```

```python
ps.stem('learning')
```

```python
ps.stem('Studied')
```

```python
from nltk import LancasterStemmer
ls=LancasterStemmer()
```

```python
ls.stem('Studied')
```

```python
ls.stem('playing')
```

```python
ls.stem('went')
```

### 3. Lemmatization :- is a process of reducing the suffixes of a word to it's root form using its meaning/dictionary word.

```python
from nltk import WordNetLemmatizer
wl=WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
wl.lemmatize('went','v')
```

```python
wl.lemmatize('running','v')
```

```python
wl.lemmatize('happier','a')
```

### 5. POS Tags (Parts of Speech Tags) :- Labelling the words with their grammatical roles

```python
from nltk import pos_tag
```

```python
sent="NLP is amazing"
sent
```

```python
txt=word_tokenize(sent)
txt
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
tags=nltk.pos_tag(txt)
tags
```

### 6. ne-chunk: means Chunking or identifying the places, persons or organizations.

```python
text="Sundar Pichai is the CEO of Google situated at New York."
```

```python
word=word_tokenize(text)
word
```

```python
txt_tag=nltk.pos_tag(word)
txt_tag
```

```python
nltk.download('maxent_ne_chunker_tab')
```

```python
nltk.download('words')
```

```python
from nltk import ne_chunk
chunk=ne_chunk(txt_tag)
chunk
```

### 7. Vectorizer :- It is a process of converting the words into raw numbers.
1. count vectorizer
2. tfidf vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
```

```python
data=["I love singing and Painting"]
data
```

```python
x=cv.fit_transform(data)
x
```

```python
print(x)
```

```python
cv.vocabulary_
```

```python
print(x.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
```

```python
data=["I Love Cricket","I Love Football","Football is Popular"]
data
```

```python
x=tfidf.fit_transform(data)
print(x)
```

```python
tfidf.vocabulary_
```
