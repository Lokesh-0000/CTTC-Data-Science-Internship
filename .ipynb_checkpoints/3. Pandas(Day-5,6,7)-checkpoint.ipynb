```python
stu={'Name':['A','B','C','D','E','F','G','H','I','J'],
     'Roll_no':[11,12,13,14,15,16,17,18,19,np.nan],
    'Course':['AIML','DA','PYTHON','WEB_DESIGN','GENAI','DSA','IOT','CLOUD','CYBER_SECURITY','JAVA']}
stu
```

```python
#dtypes:use to see the datatype of every column
df.dtypes
```

```python
#isnull():Use to see the null values present in the dataset
#null value are denoted with 'nan'
df.isnull().sum()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

### ***Titanic Dataset***

```python
titanic=pd.read_csv(r"C:\Users\KIIT\Downloads\titanic dataset.csv")
titanic
```

```python
titanic.head()
```

```python
titanic.tail()
```

```python
titanic.isnull().sum()
```

```python
titanic.dtypes
```

```python
#drop():use to delete the column
titanic=titanic.drop('Cabin',axis=1)
titanic
```

```python
d=titanic.describe(include='all')
d
```

```python
#fillna():use to fill the missing value
#Old:-     t=titanic.Age.fillna(titanic.Age.mean(),inplace=True)
titanic['Age']=titanic['Age'].fillna(titanic.Age.mean())
```

```python
titanic.isnull().sum()
```

```python
#dropna():use to delete the entire row that having the null value
titanic=titanic.dropna()
titanic
```
