# Seaborn: Statistical Data visualization

## Nature of Data
##### -->continuous data(eg.->height,weight,income,temperature,etc.)
##### -->categorical data(eg.->gender,binary value(0,1),boolean(True,False))

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sb.__version__
```

```python
auto=pd.read_csv(r"C:\Users\KIIT\Downloads\auto-mpg.csv")
auto
```

```python
#scatterplot
sns.scatterplot(x=auto.weight,y=auto.mpg)
plt.show()
```

```python
#reg plt(regression plot)
sns.regplot(x=auto.weight,y=auto.mpg)
plt.show()
```

```python
auto.origin.unique()
#1-USA
#2-Japan
#3-Europe
```

```python
#barplot
sns.barplot(x=auto.origin,y=auto.mpg)
plt.show()
```

```python
auto=auto.rename(columns={'model year':'model_year'})
auto=auto.rename(columns={'car name':'carname'})
auto
```

```python
auto.model_year.unique()
```

```python
#lineplot
sns.lineplot(x=auto.model_year,y=auto.mpg)
plt.show()
```

```python
sns.barplot(x=auto.carname,y=auto.mpg)
plt.plot()
```

```python
top=auto.sort_values(by='mpg',ascending=False).head(10)
top
```

```python
sns.barplot(x=top.mpg,y=top.carname)
plt.plot()
```

```python
#box plot
sns.boxplot(x=auto.cylinders,y=auto.mpg)
plt.show()
```

```python
#boxen plot
sns.boxenplot(x=auto.cylinders,y=auto.mpg)
plt.show()
```

```python
#displot():use to see the distribution in the data
sns.displot(auto.weight)
plt.show()
```

```python
#kde plot(kernel density estimation)
sns.kdeplot(auto.weight,fill=True,hatch='/')
plt.show()
```

```python
#countplot():use to see thefrequency
sns.countplot(x=auto.origin)
plt.show()
```

```python
sns.countplot(x=auto.cylinders)
plt.show()
```

```python
#jointplot(scatterplot,regplot,displot)
sns.jointplot(x=auto.weight,y=auto.mpg,kind='reg')
plt.show()
```

```python
#pairplot(all the possible plots)
sns.pairplot(auto)
plt.show()
```

```python
#heatmap(use to see the corelation between the columns)
c=auto.corr()
c
```

```python
auto=auto.drop('horsepower',axis=1)
auto=auto.drop('carname',axis=1)
auto
```

```python
c=auto.corr()
c
```

```python
sns.heatmap(c,annot=True,cmap='hot')
plt.show()
```
