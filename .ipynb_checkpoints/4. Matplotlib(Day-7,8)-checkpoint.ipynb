### Matpltlib:Data Visualization tool

    It is use to visualize the data.

```python
import matplotlib.pyplot as plt
```

```python
#lineplot
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.plot(x,y,c='k')
plt.title('Line Plot')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.grid() #use to see the grid lines
plt.show()
```

```python
#bar plot
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.bar(x,y,color=['r','g','b','k','m'],width=0.5)
plt.annotate('Happy',(0.6,2.3),color='orange')
plt.show()
```

```python
#Stacked bar graph
x=[1,3,5,7,9]
y=[2,4,6,8,10]

x1=[1,3,5,7,9]
y1=[2,4,6,8,10]

plt.bar(x,y)
plt.bar(x1,y1,width=0.2)
plt.show()
```

```python
#Scatter plot:-shows the relationship between numerical values
x=[1,3,5,7,9]
y=[2,4,6,8,10]
plt.scatter(x,y,s=700,marker='^')     #s:-size
plt.show()
```

```python
#pie chart
activities=['Sleeping','Playing','Food','Study','Phone']
slices=[8,4,2,5,9]
plt.pie(slices,labels=activities,
       colors=['g','m','b','cyan','r'],autopct="%1.1f%%",
       explode=(0,0,0.5,0,0))#to make a slice come out
plt.show()
```
