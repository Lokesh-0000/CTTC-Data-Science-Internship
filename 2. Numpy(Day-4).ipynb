```python
#we can join arrays in two ways 
#1->through columns
#2->through rows
a=np.random.randint(1,50,(2,2))
b=np.random.randint(1,50,(2,4))
c=np.random.randint(1,50,(4,4))
print(a)
print(b)
print(c)
```

```python
#concatenating the arrays along with the columns
#the row size must be same
#axix=1 represent the column
np.concatenate((a,b),axis=1)
```

```python
#concatenating the arrays along with the columns
#the row size must be same
#axix=0 represent the row
np.concatenate((b,c),axis=0)
```

```python
#Stacking of two arrays
#1->hstack--Horizontal Stack(joining along with column)
#2->vstack--Vertical Stack(joining along with rows)
np.hstack((a,b))
```

```python
np.vstack((b,c))
```

```python
#reshape():Changing the dimension of the array without changing the values
a=np.random.randint(1,50,size=10)
print(a)
print('Dimension:',a.ndim)
print('Shape:',a.shape)
```

```python
x=a.reshape(2,5)
print(x)
print('Dimension:',x.ndim)
print('Shape:',x.shape)
```

```python
#flatten():convert a n-dimensional array to 1-D array
y=np.random.randint(1,500,(50,50))
print(y)
print('Dimension:',y.ndim)
print('Shape:',y.shape)
```

```python
y1=y.flatten()
print(y1)
print('Dimension:',y1.ndim)
```
