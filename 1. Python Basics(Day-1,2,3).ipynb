```python
ask=input("Enter the string:")
d=a=s=0
for i in ask:
    if i.isdigit():
        d+=1
    elif i.isalpha():
        a+=1
    else:
        s+=1
print("Digits:",d,"Alphabets:",a,"Special Characters:",s)
```

```python
ask=input("Enter the string:")
b='AEIOUaeiou'
v=c=0
for i in ask:
    if i in b:
        v+=1
    else:
        c+=1
print("Vowel:",v,"Consonants:",c)
```
