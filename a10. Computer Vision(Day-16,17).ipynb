```python
#face detection through ai
face=r"C:\Users\KIIT\Downloads\haarcascade_frontalface_default.xml"
model=cv2.CascadeClassifier(face)
img_path = r"C:\Users\KIIT\Downloads\Mike-Winslow-Photography-Headshot-Portrait-3-731x1024.jpg"
img = cv2.imread(img_path)
facepoints=model.detectMultiScale(img)
for x,y,w,h in facepoints:
    cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),7)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
#Face Detection through webcam
face=r"C:\Users\KIIT\Downloads\haarcascade_frontalface_default.xml"
cap=cv2.VideoCapture(0)

ret,frame=cap.read()
while ret:
    ret,frame=cap.read()
    
    if not ret:
        break
    model=cv2.CascadeClassifier(face)
    facepoints=model.detectMultiScale(frame)
    
    for x,y,w,h in facepoints:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),7)
    cv2.imshow('Face',frame)
    if cv2.waitKey(10)==ord('q'):
        break
cv2.destroyAllWindows()
cap.release()
```
