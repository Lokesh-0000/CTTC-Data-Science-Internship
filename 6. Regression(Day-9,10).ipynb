# SVR(Support Vector Regressor)
##### Support Vector (SVR) is a machine learning tool used to predict number based on past data. BY using kernels, SVR can model highly curved, complicated patterns that simple linear models would miss.

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
y_train=sc.fit_transform(y_train.values.reshape(-1,1)).ravel() #-1 can find the length of the rows so (-1,1)can mathe array 2D
```

```python
from sklearn.svm import SVR
svr=SVR()
svr.fit(x_train,y_train)
```

```python
pred=svr.predict(x_test)
pred
```

```python
y_pred=sc.inverse_transform(pred.reshape(-1,1))
y_pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error
print('Accuracy---->',r2_score(y_test,y_pred))
print('Mean_Squared_Error---->',mean_squared_error(y_test,y_pred))
```
